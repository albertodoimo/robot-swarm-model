```python
import numpy as np
import pandas as pd
import seaborn as sb
from datetime import datetime
```

```python
control_df = pd.read_csv("control_14_08.csv")
treatment_df = pd.read_csv("treatment_14_08.csv")
```

```python
# print(len(control_df["t"][0]))
datetime.strptime(control_df['t'][1], "%Y-%m-%d %H:%M:%S.%f")
# control_df['t'][10001][-2:]

store_time_control = []
for time in control_df["t"]:
    # print(len(time), time)
    if len(time)==22:
        store_time_control.append(float(time[-5:]))
    elif len(time)==21:
        store_time_control.append(float(time[-4:]))
    else: 
        store_time_control.append(float(time[-2:]))
# store_time

store_time_treatment = []
for time in treatment_df["t"]:
    # print(len(time), time)
    if len(time)==22:
        store_time_treatment.append(float(time[-5:]))
    elif len(time)==21:
        store_time_treatment.append(float(time[-4:]))
    else: 
        store_time_treatment.append(float(time[-2:]))
```

```python
control_df["datetime_t"] = store_time_control
treatment_df["datetime_t"] = store_time_treatment
```

```python
import matplotlib.pyplot as plt
for id in np.unique(control_df["id"]):
    groupby_id = control_df.groupby("id").get_group(id)
    plt.plot(groupby_id["datetime_t"], groupby_id["nnd"], color= "r")
```

```python

for id in np.unique(treatment_df["id"]):
    groupby_id = treatment_df.groupby("id").get_group(id)
    plt.plot(groupby_id["datetime_t"], groupby_id["nnd"])

```
